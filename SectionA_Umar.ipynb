### Q1. `is` vs `==`

`==` compares values; `is` checks whether two names refer to the same object. For `a = [1, 2]` and `b = [1, 2]`, `a == b` is `True`, while `a is b` is `False`.

### Q2. Mutable and immutable types

Mutable: `list`, `set`, `dict`. Immutable: `tuple`, `str`, `int`, `float`, `bool`, `frozenset`. Immutability prevents in-place changes; dictionary keys must be hashable, and a tuple is hashable only when all its elements are hashable.

### Q3. Floating-point comparison

`print(0.1 + 0.2 == 0.3)` prints `False`: binary floating point cannot represent these decimal fractions exactly, so the sum is `0.30000000000000004`. Compare with a tolerance when approximate values are intended.

### Q4. When to choose a tuple

Use a tuple for a hashable dictionary key such as a coordinate, for a fixed record such as an RGB value, or to return several values from a function. A tuple prevents changing its own elements, though a mutable object inside it can still change.

### Q5. Dictionary lookup vs list search

Without the technical term: a dictionary usually jumps to a key’s location, while `in` on a list may compare items one by one. With a hash function: Python maps a key to a bucket, giving average `O(1)` lookup; collisions can slow it, while list membership is `O(n)`.

### Q6. `range()` vs `numpy.arange()`

`range()` is built in, produces integers, and generates values lazily. `numpy.arange()` returns a NumPy array, supports steps such as `0.25`, and supports vectorized arithmetic; floating-point steps can have precision limits.

### Q7. `append`, `extend`, and `insert`

For `a = [1, 2, 3]` and `x = [4, 5]`: `append(x)` gives `[1, 2, 3, [4, 5]]`; `extend(x)` gives `[1, 2, 3, 4, 5]`. `insert(1, x)` gives `[1, [4, 5], 2, 3]`.

### Q8. Starred unpacking

`a, *b, c = [10, 20, 30, 40, 50]` sets `a = 10`, `b = [20, 30, 40]`, and `c = 50`. With `[10, 20]`, `b` is `[]`; with `[10]`, Python raises `ValueError` because both `a` and `c` need values.

### Q9. Arithmetic operators for `(-7, 2)`

`/` divides: `-7 / 2 = -3.5`; `//` floors: `-7 // 2 = -4`; `%` gives the remainder consistent with floor division: `-7 % 2 = 1`. `**` raises to a power: `(-7) ** 2 = 49`.

### Q10. Why NumPy is faster for numerical work

NumPy stores same-type values compactly in contiguous memory, reducing storage overhead. Its vectorized operations run in compiled code and avoid much of Python’s per-item loop and type-checking overhead.
