### Q31. Armstrong numbers 1–9999

```python
armstrong_numbers = []
for num in range(1, 10000):
    # count digits
    temp = num
    digits = 0
    while temp > 0:
        digits += 1
        temp //= 10
    # sum of digits ** digits
    temp = num
    total = 0
    while temp > 0:
        total += (temp % 10) ** digits
        temp //= 10
    if total == num:
        armstrong_numbers.append(num)

print(", ".join(str(n) for n in armstrong_numbers))
print("Count:", len(armstrong_numbers))
```

### Q33. Optimized prime check + sum of primes up to N

```python
def is_prime(n):
    if n < 2:
        return False
    i = 2
    while i * i <= n:          # only go up to sqrt(n)
        if n % i == 0:
            return False
        i += 1
    return True

N = int(get("N: ", 30))
print(N, "is prime" if is_prime(N) else "is not prime")

prime_sum = 0
for num in range(2, N + 1):
    if is_prime(num):
        prime_sum += num
print(f"Sum of primes from 2 to {N}:", prime_sum)
```

### Q38. Word Frequency Analyzer

```python
review = get("Review: ", "Food was great and the food was hot. Great packaging, great delivery!")

# lowercase and keep only letters/digits/spaces
cleaned = ""
for ch in review.lower():
    if ch.isalnum() or ch == " ":
        cleaned += ch

freq = {}
for word in cleaned.split():
    freq[word] = freq.get(word, 0) + 1

# sort (word, count) pairs by count descending (simple bubble sort, no sorted()/lambda)
items = list(freq.items())
for i in range(len(items)):
    for j in range(len(items) - 1 - i):
        if items[j][1] < items[j + 1][1]:
            items[j], items[j + 1] = items[j + 1], items[j]

for word, count in items:
    print(f"{word:<12}{count}")
print("Most frequent word:", items[0][0], "| count:", items[0][1])
```

### Q45. Caesar Cipher

```python
def caesar_encrypt(message, shift):
    shift = shift % 26
    result = ""
    for ch in message:
        if "A" <= ch <= "Z":
            result += chr((ord(ch) - ord("A") + shift) % 26 + ord("A"))
        elif "a" <= ch <= "z":
            result += chr((ord(ch) - ord("a") + shift) % 26 + ord("a"))
        else:
            result += ch               # digits, spaces, symbols unchanged
    return result

message = get("Message: ", "Hello, India!")
shift = int(get("Shift: ", 3))
print(caesar_encrypt(message, shift))
```

## C7 — Functions (Q46–Q47)
### Q46. `calculate_gst`

```python
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    amount = float(amount)
    if gst_inclusive:
        base = amount / (1 + gst_rate / 100)    # strip GST out of the total
        gst = amount - base
        total = amount
    else:
        base = amount
        gst = amount * gst_rate / 100
        total = base + gst
    return (round(base, 2), round(gst, 2), round(total, 2))

print(calculate_gst(1000))
print(calculate_gst(1180, gst_inclusive=True))
print(calculate_gst(500, gst_rate=5))
```

### Q47. Recursive `power`

```python
def power(base, exp):
    if exp == 0:
        return 1
    if exp < 0:
        return 1 / power(base, -exp)
    return base * power(base, exp - 1)

print(power(2, 10))
print(power(5, 0))
print(power(3, -2))
```
