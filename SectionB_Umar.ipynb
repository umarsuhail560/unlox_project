### Q12. Rebinding vs in-place mutation

**Predicted output**
```text
x = [1, 2, 3, 4]
y = [1, 2, 3]
a = [1, 2, 3, 4]
b = [1, 2, 3, 4]
```

**Why:** `x + [4]` creates a new list and rebinds `x`; `a += [4]` mutates the list shared by `a` and `b`.

````python
x = [1, 2, 3]
y = x
x = x + [4]
print("x =", x)
print("y =", y)

a = [1, 2, 3]
b = a
a += [4]
print("a =", a)
print("b =", b)
````

**Predicted output**


**Why:** strings concatenate or repeat, while integers add; multiplying an integer by a string repeats it. Adding a string and integer raises `TypeError`, which the notebook catches before continuing.

````python
print("10" + "20")
print(10 + 20)
print("10" * 3)
print(10 * "3")
try:
    print("10" + 20)
except TypeError as e:
    print("TypeError:", e)
print(int("10") + 20)
````

### Q15. Short-circuit evaluation

**Predicted output**
```text

5
hello
0
10
default

```

**Why:** `or` returns its first truthy operand and `and` returns its first falsy operand (or the last operand). A non-empty list such as `[0]` is truthy.

````python
print(0 or 5)
print(5 or 0)
print(None or "hello")
print(0 and 5)
print(5 and 10)
print([] or "default")
print([0] or "default")
````
