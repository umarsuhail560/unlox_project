```python
=====================================================
 WEEK 1 ASSIGNMENT - SECTION A (THEORY)
 Questions 1 to 10 - answers in simple English
 Name: Umar
=====================================================


-----------------------------------------------------
Q1. What is the difference between "is" and "=="?
-----------------------------------------------------
Think of two houses that look exactly the same: same colour, same size.

"==" asks: "Do these two look the same?"  -> Yes.
"is" asks: "Is this literally the same house?"  -> No.

In Python:
    a = [1, 2, 3]
    b = [1, 2, 3]
    a == b   -> True   (the values are the same)
    a is b   -> False  (they are two different lists stored in two different places in memory)

So "==" checks the VALUE, and "is" checks whether it is the SAME OBJECT.


-----------------------------------------------------
Q2. Which of these are mutable and which are immutable?
-----------------------------------------------------
Mutable means "can be changed after it is created".
Immutable means "cannot be changed once it is created".

Mutable   : list, set, dict
Immutable : tuple, str, int, float, bool, frozenset

Why does it matter? If something can't change, nobody can break it by
mistake. Also, only things that can't change are allowed to be dictionary
keys. It makes the code safer.


-----------------------------------------------------
Q3. What is the output of print(0.1 + 0.2 == 0.3) and why?
-----------------------------------------------------
The output is False.

Why? Computers store numbers using only 0s and 1s (binary). Some decimal
numbers like 0.1 cannot be written exactly in binary. It is just like
1/3 in normal maths: 0.3333... it never ends, so we have to cut it
somewhere. Because of this tiny cut, 0.1 + 0.2 becomes
0.30000000000000004, which is not exactly 0.3.

This is NOT a Python bug. Every programming language does the same.
The safe way is to check if the difference is very, very small:
abs(a - b) < 1e-9


-----------------------------------------------------
Q4. Give 3 situations where you would choose a tuple over a list.
-----------------------------------------------------
1) As a dictionary key. For example (12.97, 77.59) for the location of a
   city. A list is not allowed as a key, but a tuple is.

2) For data that should never change, like the days of the week or the
   RGB colour (255, 0, 0). A tuple makes sure nobody changes it by mistake.

3) When a function returns more than one value. For example
   "return base, gst, total" gives you a tuple, and you can just unpack it.


-----------------------------------------------------
Q5. Why is dictionary lookup O(1) but searching a list with "in" is O(n)?
-----------------------------------------------------
Without using the word "hash":
A dictionary is like a hostel with locker numbers. If you tell it the
name, it already knows exactly which locker to open, so it goes straight
there. A list has no such system. To find something, it has to check item 1,
then item 2, then item 3... until it finds it. If the list is big, that takes
a long time.

With the word "hash":
When you give a key to a dictionary, Python runs it through a "hash
function" which turns the key into a number. That number tells Python the
exact position where the value is stored, so it jumps straight there. That
is O(1), meaning almost the same time for any size. For a list, "in" compares
every item one by one, so more items means more time. That is O(n).


-----------------------------------------------------
Q6. Difference between range() and numpy.arange() (3 differences)
-----------------------------------------------------
1) Floats: range() works only with whole numbers. arange() also works with
   decimals, for example np.arange(0, 1, 0.25).

2) What you get back: range() gives a small "range object" that makes
   numbers only when needed. arange() gives a real NumPy array that is
   already stored in memory.

3) Maths: with arange() you can do maths on the whole thing at once,
   like np.arange(5) * 2. With range() you can't.


-----------------------------------------------------
Q7. append vs extend vs insert  (a = [1, 2, 3] and x = [4, 5])
-----------------------------------------------------
append(x)    -> adds x as ONE single item at the end.
                a becomes [1, 2, 3, [4, 5]]

extend(x)    -> opens x and adds each item one by one at the end.
                a becomes [1, 2, 3, 4, 5]

insert(1, x) -> puts x as ONE item at the position you choose (here, position 1).
                a becomes [1, [4, 5], 2, 3]


-----------------------------------------------------
Q8. What does  a, *b, c = [10, 20, 30, 40, 50]  do?
-----------------------------------------------------
The first value goes to a, the last value goes to c, and the "*" tells
Python to put everything in the middle into b as a list.

    a = 10
    b = [20, 30, 40]
    c = 50

With [10, 20]: a = 10, c = 20, and b is an empty list [] because nothing is
left in the middle.

With [10]: it gives an ERROR (ValueError), because there are two names that
need a value (a and c) but only one value is available.


-----------------------------------------------------
Q9. Difference between  /   //   %   **   for (-7, 2)
-----------------------------------------------------
/   normal division, always gives a decimal.      -7 / 2  = -3.5
//  floor division, rounds DOWN to the lower whole number.
                                                  -7 // 2 = -4
    (careful: it goes down to -4, not up to -3)
%   remainder after dividing.                     -7 % 2  = 1
**  power.                                        (-7) ** 2 = 49


-----------------------------------------------------
Q10. Why is NumPy much faster than Python lists? (3 reasons)
-----------------------------------------------------
1) Memory: a NumPy array keeps all the numbers side by side in one
   continuous block of memory. A list only keeps addresses that point to
   numbers lying all over the memory, so the computer has to jump around.

2) Speed of the loop: NumPy does its work in C code behind the scenes, and
   it works on the whole array at once. A Python for-loop handles one item
   at a time and is slow.

3) Same type: every item in a NumPy array is the same type (like all
   integers), so Python doesn't need to check the type of each item every time.

=====================================================
                    END OF SECTION A
=====================================================
```
